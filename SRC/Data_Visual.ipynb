```python
from pathlib import Path
import pandas as pd

# Notebook is inside SRC → go one level up to project root
BASE = Path.cwd().parents[0]

CLEANED = BASE / "Cleaned"

print("Project base:", BASE)
print("Cleaned folder exists:", CLEANED.exists())

```

```python
sales = pd.read_csv(CLEANED / "sales_clean.csv", parse_dates=["Date"])
features = pd.read_csv(CLEANED / "features_clean.csv", parse_dates=["Date"])
stores = pd.read_csv(CLEANED / "stores_clean.csv")

print("Sales rows:", len(sales))
print("Features rows:", len(features))
print("Stores rows:", len(stores))
```

```python
df = sales.merge(
    features,
    on=["Store_ID", "Date"],
    how="left"
)

print("Rows after sales + features merge:", len(df))
```

```python
df = df.merge(
    stores,
    on="Store_ID",
    how="left"
)

print("Rows after adding stores:", len(df))
```

```python
print("Rows before merge:", len(sales))
print("Rows after merge :", len(df))

if len(df) != len(sales):
    raise ValueError("❌ Row count mismatch after merge!")

print("Missing Store_Type:", df["Store_Type"].isna().sum())
print("Missing Temperature:", df["Temperature"].isna().sum())
```

```python
df["Is_Return_Week"] = (df["Weekly_Sales"] < 0).astype(int)

df[["Weekly_Sales", "Is_Return_Week"]].head()
```

```python
output_path = CLEANED / "retail_analytics.csv"
df.to_csv(output_path, index=False)

print("✅ Step 4 complete")
print("Final dataset saved as:", output_path)
print("Total columns:", len(df.columns))
```
